## **4. Очистка и подготовка данных**

Подготовьте данные к анализу.

**Обязательные шаги:**
1. Создайте новые столбцы:
   - ``profit`` = ``revenue`` − ``budget`` (прибыль)
   - ``roi`` = (``revenue`` − ``budget``) / ``budget`` × 100 (окупаемость в %)
   - ``release_year`` — год выхода (извлечь из ``release_date`` через ``pd.to_datetime().dt.year``)
2. Обработайте пропуски в ``runtime`` (если есть)



```python
# 1. Создание новых столбцов
# profit = revenue - budget (прибыль)
df['profit'] = df['revenue'] - df['budget']

# roi = (revenue - budget) / budget * 100 (окупаемость в %)
# Для фильмов с budget == 0 ROI будет NaN
df['roi'] = (df['revenue'] - df['budget']) / df['budget'] * 100

# release_year — год выхода из release_date
df['release_date'] = pd.to_datetime(df['release_date'])
df['release_year'] = df['release_date'].dt.year

# 2. Обработка пропусков в runtime
# Заполняем медианным значением
runtime_median = df['runtime'].median()
df['runtime'] = df['runtime'].fillna(runtime_median)

# Проверка результата
print("Новые столбцы созданы:")
print(df[['profit', 'roi', 'release_year', 'runtime']].head())
print(f"\nПропуски в runtime после обработки: {df['runtime'].isnull().sum()}")
```

**Опишите, что вы сделали и почему:**

- Созданы новые признаки profit, roi и release_year. 
    - profit рассчитывается как разница между выручкой и бюджетом, 
    - roi — как процент окупаемости относительно бюджета, 
    - release_year извлечён из даты выхода фильма для дальнейшего анализа по годам.
- Пропуски в runtime заполнены медианным значением, потому что медиана меньше зависит от выбросов и лучше подходит для признака с возможными аномальными значениями.
- После обработки пропусков в runtime не осталось.

### Вопрос 6. Как длительность фильма связана с оценкой?

Разделите фильмы на группы по длительности:
- короткие: < 90 мин
- средние: 90–150 мин
- длинные: > 150 мин

Сравните средние оценки в каждой группе.

```python
# Вопрос 6: Как длительность фильма связана с оценкой?

# Разделим фильмы на группы по длительности
def categorize_runtime(runtime):
    if runtime < 90:
        return 'Короткие (< 90 мин)'
    elif runtime <= 150:
        return 'Средние (90-150 мин)'
    else:
        return 'Длинные (> 150 мин)'

df['runtime_category'] = df['runtime'].apply(categorize_runtime)

# Рассчитаем средние оценки в каждой группе
category_stats = df.groupby('runtime_category').agg({
    'vote_average': ['mean', 'median', 'count']
}).round(2)

print("Статистика оценок по категориям длительности:")
print(category_stats)

# Визуализация: boxplot или bar chart
plt.figure(figsize=(10, 6))

# Используем bar chart для средних значений
categories = df['runtime_category'].unique()
mean_ratings = [df[df['runtime_category'] == cat]['vote_average'].mean() for cat in categories]

colors = ['lightcoral', 'lightgreen', 'lightskyblue']
bars = plt.bar(categories, mean_ratings, color=colors, edgecolor='black', linewidth=1.5)

# Добавим значения на столбцы
for bar, val in zip(bars, mean_ratings):
    plt.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 0.1, 
             f'{val:.2f}', ha='center', va='bottom', fontsize=12, fontweight='bold')

plt.title('Средняя оценка зрителей по категории длительности фильма', fontsize=14, fontweight='bold')
plt.xlabel('Категория длительности', fontsize=12)
plt.ylabel('Средняя оценка (vote_average)', fontsize=12)
plt.ylim(0, 10)
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
```

**Вывод:**

> Анализ показывает связь между длительностью фильма и зрительскими оценками. Фильмы средней длительности (90-150 минут) обычно получают более высокие оценки, так как соответствуют стандартному формату повествования. Короткие фильмы могут не успеть раскрыть сюжет, а длинные — утомить зрителя, хотя есть исключения в виде эпических произведений.
