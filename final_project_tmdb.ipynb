## **4. Очистка и подготовка данных**

Подготовьте данные к анализу.

**Обязательные шаги:**
1. Создайте новые столбцы:
   - ``profit`` = ``revenue`` − ``budget`` (прибыль)
   - ``roi`` = (``revenue`` − ``budget``) / ``budget`` × 100 (окупаемость в %)
   - ``release_year`` — год выхода (извлечь из ``release_date`` через ``pd.to_datetime().dt.year``)
2. Обработайте пропуски в ``runtime`` (если есть)



```python
# 1. Создание новых столбцов
# profit = revenue - budget (прибыль)
df['profit'] = df['revenue'] - df['budget']

# roi = (revenue - budget) / budget * 100 (окупаемость в %)
# Для фильмов с budget == 0 ROI будет NaN
df['roi'] = (df['revenue'] - df['budget']) / df['budget'] * 100

# release_year — год выхода из release_date
df['release_date'] = pd.to_datetime(df['release_date'])
df['release_year'] = df['release_date'].dt.year

# 2. Обработка пропусков в runtime
# Заполняем медианным значением
runtime_median = df['runtime'].median()
df['runtime'] = df['runtime'].fillna(runtime_median)

# Проверка результата
print("Новые столбцы созданы:")
print(df[['profit', 'roi', 'release_year', 'runtime']].head())
print(f"\nПропуски в runtime после обработки: {df['runtime'].isnull().sum()}")
```

**Опишите, что вы сделали и почему:**

- Созданы новые признаки profit, roi и release_year. 
    - profit рассчитывается как разница между выручкой и бюджетом, 
    - roi — как процент окупаемости относительно бюджета, 
    - release_year извлечён из даты выхода фильма для дальнейшего анализа по годам.
- Пропуски в runtime заполнены медианным значением, потому что медиана меньше зависит от выбросов и лучше подходит для признака с возможными аномальными значениями.
- После обработки пропусков в runtime не осталось.

## **5. Анализ распределений**

Исследуйте, как распределены ключевые числовые признаки:
- Постройте гистограммы для ``budget``, ``revenue``, ``vote_average``, ``runtime``
- На каждой гистограмме отметьте **среднее** и **медиану** вертикальными линиями (``.axvline()``)

```python
# Ваш код здесь:

# Анализ распределений ключевых числовых признаков

fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# Список признаков для анализа
features = ['budget', 'revenue', 'vote_average', 'runtime']
titles = ['Budget', 'Revenue', 'Vote Average', 'Runtime']

for idx, (col, title) in enumerate(zip(features, titles)):
    ax = axes[idx // 2, idx % 2]
    
    # Гистограмма
    sns.histplot(df[col], bins=50, kde=False, ax=ax, color='skyblue', edgecolor='black')
    
    # Среднее и медиана
    mean_val = df[col].mean()
    median_val = df[col].median()
    
    ax.axvline(mean_val, color='red', linestyle='--', linewidth=2, label=f'Среднее: {mean_val:.2f}')
    ax.axvline(median_val, color='green', linestyle='-', linewidth=2, label=f'Медиана: {median_val:.2f}')
    
    ax.set_title(f'Распределение {title}', fontsize=14)
    ax.set_xlabel(col)
    ax.set_ylabel('Частота')
    ax.legend()

plt.tight_layout()
plt.show()

# Вывод статистик для сравнения
print("=== Сравнение среднего и медианы ===")
print(f"{'Признак':<15} {'Среднее':>15} {'Медиана':>15} {'Разница':>15}")
print("-" * 60)
for col in features:
    mean_val = df[col].mean()
    median_val = df[col].median()
    diff = abs(mean_val - median_val)
    print(f"{col:<15} {mean_val:>15.2f} {median_val:>15.2f} {diff:>15.2f}")

# Boxplot для поиска выбросов
plt.figure(figsize=(14, 6))
sns.boxplot(data=df[features])
plt.title('Boxplot для поиска выбросов', fontsize=14)
plt.ylabel('Значение')
plt.xlabel('Признак')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

### Вопрос 1. Какие жанры представлены чаще всего?

Постройте столбчатую диаграмму (bar chart) частот жанров (``primary_genre``).

```python
# Ваш код здесь
# Добавим признак primary_genre для анализа жанров

def get_primary_genre(genres_str):
    try:
        genres_list = json.loads(str(genres_str).replace("'", '"'))
        if len(genres_list) > 0:
            return genres_list[0]['name']
        else:
            return 'Unknown'
    except:
        return 'Unknown'

df['primary_genre'] = df['genres'].apply(get_primary_genre)

print("Признак primary_genre создан!")
print(f"Примеры: {df['primary_genre'].head(10).tolist()}")

# Вопрос 1: Какие жанры представлены чаще всего?

# Подсчитаем частоту каждого первичного жанра
genre_counts = df['primary_genre'].value_counts().sort_values(ascending=False)

# Построим столбчатую диаграмму
plt.figure(figsize=(12, 6))
plt.bar(genre_counts.index, genre_counts.values, color='steelblue', edgecolor='black')
plt.title('Частота жанров в датасете (по первичному жанру)', fontsize=14, fontweight='bold')
plt.xlabel('Жанр', fontsize=12)
plt.ylabel('Количество фильмов', fontsize=12)
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()

print("Топ-5 самых частых жанров:")
print(genre_counts.head())
```

**Вывод:**

На основе анализа видно, какие жанры наиболее популярны среди filmmakers. Чаще всего встречаются жанры Drama, Action и Comedy, что отражает коммерческую ориентацию киноиндустрии на массового зрителя.

### Вопрос 2. Какие жанры приносят наибольшие средние сборы?

Рассчитайте средний ``revenue`` по жанрам (через ``.groupby()`` + ``.mean()``) и визуализируйте результат.

### Вопрос 3. Как менялись сборы и бюджеты по годам?

Сгруппируйте данные по ``release_year`` через ``.groupby()`` + ``.agg()`` и постройте линейный график динамики среднего бюджета и средних сборов.

```python
# Ваш код здесь
# Вопрос 3: Как менялись сборы и бюджеты по годам?

# Сгруппируем данные по годам и рассчитаем средние значения
yearly_stats = df.groupby('release_year').agg({
    'budget': 'mean',
    'revenue': 'mean'
}).reset_index()

# Построим линейный график динамики
plt.figure(figsize=(14, 7))
plt.plot(yearly_stats['release_year'], yearly_stats['budget'] / 1_000_000, 
         marker='o', linewidth=2, label='Средний бюджет (млн USD)', color='red')
plt.plot(yearly_stats['release_year'], yearly_stats['revenue'] / 1_000_000, 
         marker='s', linewidth=2, label='Средние сборы (млн USD)', color='blue')

plt.title('Динамика средних бюджетов и кассовых сборов по годам', fontsize=14, fontweight='bold')
plt.xlabel('Год выхода', fontsize=12)
plt.ylabel('Сумма (млн USD)', fontsize=12)
plt.legend(fontsize=10)
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()

print("Статистика по годам (первые 10 лет):")
print(yearly_stats.head(10))
```

**Вывод:**

Наблюдается рост как средних бюджетов, так и кассовых сборов с течением времени. Особенно заметен рост после 2010 года.

### Вопрос 4. Какие фильмы имеют наибольший ROI (окупаемость)?

Найдите топ-10 фильмов с наибольшим ROI (через `.sort_values()`). Какой жанр преобладает?

💡 *Подсказка:* если ROI имеет аномально большие значения, подумайте — почему?

```python
# Ваш код здесь
# Вопрос 4: Какие фильмы имеют наибольший ROI (окупаемость)?

# Исключаем фильмы с отсутствующим или нереалистично малым бюджетом
df_positive_budget = df[
    (df['budget'] >= 10000) &
    (df['revenue'] > 0)
].copy()

# Топ-10 фильмов по ROI
top_roi = df_positive_budget.nlargest(
    10, 'roi'
)[['title', 'budget', 'revenue', 'roi', 'primary_genre']]

print("Топ-10 фильмов по ROI (окупаемости):")
print(top_roi.to_string(index=False))

# Визуализация
plt.figure(figsize=(12, 6))
plt.barh(top_roi['title'], top_roi['roi'])

plt.title(
    'Топ-10 фильмов по окупаемости (ROI %)',
    fontsize=14,
    fontweight='bold'
)
plt.xlabel('ROI (%)', fontsize=12)
plt.ylabel('Фильм', fontsize=12)

plt.gca().invert_yaxis()
plt.grid(axis='x', alpha=0.3)
plt.tight_layout()
plt.show()
```

**Вывод:**

Фильмы с аномально низкими значениями бюджета были исключены, поскольку они приводят к искусственно завышенному ROI.

## **7. Корреляционная матрица**

Постройте корреляционную матрицу для числовых признаков:
``budget``, ``revenue``, ``vote_average``, ``vote_count``, ``runtime``, ``popularity``.

- Рассчитайте коэффициент корреляции, обоснуйте выбор подхода
- Визуализируйте через ``sns.heatmap()`` с аннотациями (``annot=True``)
- Какие пары признаков имеют **сильную** корреляцию (|r| > 0.7)?
- Какие признаки сильнее всего коррелируют с ``revenue``?


```python
# Выбираем числовые признаки для анализа
numeric_features = ['budget', 'revenue', 'vote_average', 'vote_count', 'runtime', 'popularity']

# Создаём подмножество данных с выбранными признаками
df_numeric = df[numeric_features].copy()

# Обоснование выбора коэффициента корреляции Пирсона:
# - Признаки имеют непрерывное распределение
# - Нас интересуют линейные взаимосвязи между переменными
# - Коэффициент Пирсона подходит для оценки силы и направления линейной связи

# Рассчитываем корреляционную матрицу (метод Пирсона по умолчанию)
corr_matrix = df_numeric.corr(method='pearson')

print("=== Корреляционная матрица ===")
print(corr_matrix.round(3))

# Визуализация корреляционной матрицы через heatmap
plt.figure(figsize=(10, 8))
sns.heatmap(corr_matrix, annot=True, fmt='.3f', cmap='coolwarm', center=0, 
            linewidths=0.5, square=True, cbar_kws={"shrink": 0.8})
plt.title('Корреляционная матрица числовых признаков', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()

# Анализ сильных корреляций (|r| > 0.7)
print("\n=== Пары признаков с сильной корреляцией (|r| > 0.7) ===")
strong_correlations = []

for i in range(len(corr_matrix.columns)):
    for j in range(i + 1, len(corr_matrix.columns)):
        corr_value = corr_matrix.iloc[i, j]
        if abs(corr_value) > 0.7:
            feat1 = corr_matrix.columns[i]
            feat2 = corr_matrix.columns[j]
            strong_correlations.append((feat1, feat2, corr_value))
            print(f"{feat1} ↔ {feat2}: r = {corr_value:.3f}")

if not strong_correlations:
    print("Нет пар с корреляцией |r| > 0.7")

# Определение признаков, наиболее сильно коррелирующих с revenue
print("\n=== Корреляция с revenue (по убыванию абсолютного значения) ===")
revenue_corr = corr_matrix['revenue'].drop('revenue').sort_values(key=abs, ascending=False)

for feature, corr_value in revenue_corr.items():
    print(f"{feature}: r = {corr_value:.3f}")

# Вывод самого сильного коррелирующего признака
strongest_corr_feature = revenue_corr.abs().idxmax()
strongest_corr_value = revenue_corr.max() if revenue_corr.max() > abs(revenue_corr.min()) else revenue_corr.min()
print(f"\nСильнее всего с revenue коррелирует '{strongest_corr_feature}' (r = {strongest_corr_value:.3f})")
```

**Интерпретация корреляционной матрицы:**

В корреляционной матрице наблюдаются несколько сильных положительных связей между признаками.

- Самая сильная корреляция выявлена между выручкой (revenue) и количеством голосов (vote_count) — r = 0.781. Это может означать, что фильмы с высокой выручкой обычно привлекают больше зрителей и, соответственно, получают больше оценок на платформе.
- Также сильная положительная связь наблюдается между количеством голосов (vote_count) и популярностью (popularity) — r = 0.778. Чем популярнее фильм, тем больше пользователей проявляют к нему интерес и оставляют оценки.
- Между бюджетом (budget) и выручкой (revenue) также имеется сильная положительная корреляция — r = 0.731. Это может быть связано с тем, что фильмы с большим бюджетом имеют больше возможностей для качественного производства, рекламы и широкого проката, что способствует увеличению кассовых сборов.
- Умеренная связь наблюдается между выручкой и популярностью — r = 0.645. При этом продолжительность фильма (runtime) и средняя оценка (vote_average) имеют сравнительно слабую связь с выручкой — r = 0.251 и r = 0.197 соответственно.

Таким образом, среди рассмотренных признаков наиболее тесно с выручкой связаны количество голосов, бюджет и популярность фильма. Все выявленные связи положительные: при увеличении одного показателя второй в среднем также увеличивается. При этом корреляция показывает наличие статистической связи, но сама по себе не доказывает причинно-следственную зависимость.

## **9. Доверительный интервал для средней оценки**

Постройте **95% доверительный интервал** для среднего значения ``vote_average`` всех фильмов в очищенном датасете.

**Шаги:**
1. Рассчитайте выборочное среднее и стандартное отклонение (``.std(ddof=1)``)
2. Рассчитайте стандартную ошибку среднего
3. Найдите критическое значение t-распределения через ``stats.t.ppf()``
4. Вычислите погрешность и границы интервала
5. Проверьте результат через ``stats.t.interval()``

**Интерпретация:** что означает этот доверительный интервал для киностудии? Какую среднюю оценку можно ожидать от фильма?

```python
# Шаг 1: Рассчитаем выборочное среднее и стандартное отклонение
mean_vote = df['vote_average'].mean()
std_vote = df['vote_average'].std(ddof=1)
n = len(df['vote_average'])

print(f"Выборочное среднее: {mean_vote:.4f}")
print(f"Стандартное отклонение (ddof=1): {std_vote:.4f}")
print(f"Размер выборки: {n}")

# Шаг 2: Рассчитаем стандартную ошибку среднего
se = std_vote / np.sqrt(n)
print(f"\nСтандартная ошибка среднего: {se:.4f}")

# Шаг 3: Найдём критическое значение t-распределения для 95% ДИ
alpha = 0.05
df_degrees = n - 1  # степени свободы
t_critical = stats.t.ppf(1 - alpha/2, df_degrees)
print(f"\nКритическое значение t ({alpha*100}% ДИ, df={df_degrees}): {t_critical:.4f}")

# Шаг 4: Вычислим погрешность и границы интервала
margin_error = t_critical * se
ci_lower = mean_vote - margin_error
ci_upper = mean_vote + margin_error
print(f"\nПогрешность: {margin_error:.4f}")
print(f"95% доверительный интервал: [{ci_lower:.4f}, {ci_upper:.4f}]")

# Шаг 5: Проверим результат через stats.t.interval()
ci_check = stats.t.interval(1 - alpha, df_degrees, loc=mean_vote, scale=se)
print(f"\nПроверка через stats.t.interval(): [{ci_check[0]:.4f}, {ci_check[1]:.4f}]")
```

**Интерпретация доверительного интервала:**

С вероятностью 95% истинное среднее значение оценок (`vote_average`) всех фильмов в генеральной совокупности находится в диапазоне **[6.06, 6.12]**.

**Что это означает для киностудии:**
- Средняя оценка типичного фильма составляет около **6.09 баллов** из 10
- При планировании нового фильма можно ожидать, что его средняя зрительская оценка будет находиться в этом диапазоне
- Это базовый ориентир: если студия хочет получить фильм с рейтингом выше среднего, необходимо инвестировать в качество производства, подбор актёров и маркетинг
- Узкий доверительный интервал (благодаря большой выборке ~4800 фильмов) указывает на высокую точность оценки среднего значения

### **Гипотеза 1. Отличаются ли оценки англоязычных и неанглоязычных фильмов?**

Проверьте, отличается ли средняя оценка (``vote_average``) фильмов на английском языке (``original_language == 'en'``) от фильмов на других языках.

**Шаги:**
1. Посчитайте количество фильмов в каждой группе — достаточно ли наблюдений?
2. Постройте гистограммы обеих групп на одном графике (с ``alpha``)
3. Сформулируйте H₀ и H₁
4. Проведите двухвыборочный тест
5. Рассчитайте размер эффекта (Cohen's d) с помощью ``np.sqrt()``
6. Интерпретируйте: есть ли статистически **и практически** значимая разница?

💡 **Формула Cohen's d:**

d = (x̄₁ − x̄₂) / s_pooled,  где  s_pooled = √((s₁² + s₂²) / 2)

```python
# Гипотеза 1: Отличаются ли оценки англоязычных и неанглоязычных фильмов?

# Шаг 1: Разделение на группы
en_films = df[df['original_language'] == 'en']['vote_average']
non_en_films = df[df['original_language'] != 'en']['vote_average']

# Количество наблюдений
n_en = len(en_films)
n_non_en = len(non_en_films)
print(f"Англоязычные фильмы: {n_en}")
print(f"Неанглоязычные фильмы: {n_non_en}")

# Шаг 2: Гистограммы
plt.figure(figsize=(10, 6))
plt.hist(en_films, bins=30, alpha=0.5, label='English', color='blue')
plt.hist(non_en_films, bins=30, alpha=0.5, label='Non-English', color='red')
plt.title('Distribution of Vote Average by Language')
plt.xlabel('Vote Average')
plt.ylabel('Frequency')
plt.legend()
plt.savefig('histogram_comparison.png', dpi=150, bbox_inches='tight')
plt.show()

# Шаг 3: Формулировка гипотез
# H₀: μ_en = μ_non_en (средние оценки равны)
# H₁: μ_en ≠ μ_non_en (средние оценки различаются)

# Шаг 4: Двухвыборочный t-тест Уэлша (не предполагаем равенство дисперсий)
t_stat, p_value = stats.ttest_ind(en_films, non_en_films, equal_var=False)
print(f"\nt-статистика: {t_stat:.4f}")
print(f"p-value: {p_value:.2e}")

# Шаг 5: Cohen's d
mean_en = en_films.mean()
mean_non_en = non_en_films.mean()
std_en = en_films.std(ddof=1)
std_non_en = non_en_films.std(ddof=1)

# s_pooled = √((s₁² + s₂²) / 2)
s_pooled = np.sqrt((std_en**2 + std_non_en**2) / 2)
cohens_d = (mean_en - mean_non_en) / s_pooled

print(f"\nСреднее (en): {mean_en:.4f}")
print(f"Среднее (non-en): {mean_non_en:.4f}")
print(f"Cohen's d: {cohens_d:.4f}")

# Шаг 6: Интерпретация
alpha = 0.05
print(f"\n=== ИНТЕРПРЕТАЦИЯ ===")
print(f"H₀: Средние оценки англоязычных и неанглоязычных фильмов равны")
print(f"H₁: Средние оценки англоязычных и неанглоязычных фильмов различаются")
print(f"\nУровень значимости α = {alpha}")
print(f"p-value = {p_value:.2e}")

if p_value < alpha:
    print("→ Отклоняем H₀: разница статистически значима")
else:
    print("→ Не отклоняем H₀: разница не статистически значима")

# Интерпретация Cohen's d
print(f"\nCohen's d = {cohens_d:.4f}")
if abs(cohens_d) < 0.2:
    effect = "очень малый/отсутствует"
elif abs(cohens_d) < 0.5:
    effect = "малый"
elif abs(cohens_d) < 0.8:
    effect = "средний"
else:
    effect = "большой"
print(f"→ Размер эффекта: {effect}")

print(f"\nРазница в средних: {mean_en - mean_non_en:.4f} балла")
print("Вывод: Англоязычные фильмы имеют статистически значимо более высокую оценку,")
print("но размер эффекта мал (практическая значимость ограничена).")
```
