## **4. Очистка и подготовка данных**

Подготовьте данные к анализу.

**Обязательные шаги:**
1. Создайте новые столбцы:
   - ``profit`` = ``revenue`` − ``budget`` (прибыль)
   - ``roi`` = (``revenue`` − ``budget``) / ``budget`` × 100 (окупаемость в %)
   - ``release_year`` — год выхода (извлечь из ``release_date`` через ``pd.to_datetime().dt.year``)
2. Обработайте пропуски в ``runtime`` (если есть)



```python
# 1. Создание новых столбцов
# profit = revenue - budget (прибыль)
df['profit'] = df['revenue'] - df['budget']

# roi = (revenue - budget) / budget * 100 (окупаемость в %)
# Для фильмов с budget == 0 ROI будет NaN
df['roi'] = (df['revenue'] - df['budget']) / df['budget'] * 100

# release_year — год выхода из release_date
df['release_date'] = pd.to_datetime(df['release_date'])
df['release_year'] = df['release_date'].dt.year

# 2. Обработка пропусков в runtime
# Заполняем медианным значением
runtime_median = df['runtime'].median()
df['runtime'] = df['runtime'].fillna(runtime_median)

# Проверка результата
print("Новые столбцы созданы:")
print(df[['profit', 'roi', 'release_year', 'runtime']].head())
print(f"\nПропуски в runtime после обработки: {df['runtime'].isnull().sum()}")
```

**Опишите, что вы сделали и почему:**

- Созданы новые признаки profit, roi и release_year. 
    - profit рассчитывается как разница между выручкой и бюджетом, 
    - roi — как процент окупаемости относительно бюджета, 
    - release_year извлечён из даты выхода фильма для дальнейшего анализа по годам.
- Пропуски в runtime заполнены медианным значением, потому что медиана меньше зависит от выбросов и лучше подходит для признака с возможными аномальными значениями.
- После обработки пропусков в runtime не осталось.

```python
# Анализ распределений ключевых числовых признаков

fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# Список признаков для анализа
features = ['budget', 'revenue', 'vote_average', 'runtime']
titles = ['Budget', 'Revenue', 'Vote Average', 'Runtime']

for idx, (col, title) in enumerate(zip(features, titles)):
    ax = axes[idx // 2, idx % 2]
    
    # Гистограмма
    sns.histplot(df[col], bins=50, kde=False, ax=ax, color='skyblue', edgecolor='black')
    
    # Среднее и медиана
    mean_val = df[col].mean()
    median_val = df[col].median()
    
    ax.axvline(mean_val, color='red', linestyle='--', linewidth=2, label=f'Среднее: {mean_val:.2f}')
    ax.axvline(median_val, color='green', linestyle='-', linewidth=2, label=f'Медиана: {median_val:.2f}')
    
    ax.set_title(f'Распределение {title}', fontsize=14)
    ax.set_xlabel(col)
    ax.set_ylabel('Частота')
    ax.legend()

plt.tight_layout()
plt.show()

# Вывод статистик для сравнения
print("=== Сравнение среднего и медианы ===")
print(f"{'Признак':<15} {'Среднее':>15} {'Медиана':>15} {'Разница':>15}")
print("-" * 60)
for col in features:
    mean_val = df[col].mean()
    median_val = df[col].median()
    diff = abs(mean_val - median_val)
    print(f"{col:<15} {mean_val:>15.2f} {median_val:>15.2f} {diff:>15.2f}")

# Boxplot для поиска выбросов
plt.figure(figsize=(14, 6))
sns.boxplot(data=df[features])
plt.title('Boxplot для поиска выбросов', fontsize=14)
plt.ylabel('Значение')
plt.xlabel('Признак')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
